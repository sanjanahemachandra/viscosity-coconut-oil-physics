```python
import pandas as pd

# Constants
g = 9.81  # Acceleration due to gravity (m/s^2)
RHO_FLUID = 900  # Density of coconut oil (kg/m^3)
RHO_AIR = 1.4  # Density of air (kg/m^3)

# Load the Excel file
tracker = pd.ExcelFile('tracker_data.xlsx')

# Prepare a dictionary to store terminal velocities for each sheet
terminal_velocities = {}

# Process each sheet in the Excel file
for sheet in tracker.sheet_names:
    # Load data from the sheet
    data = tracker.parse(sheet)
    
    # Extract time and velocity columns
    time = data["t"]  # Replace 't' with the actual time column name
    velocity = data["x"]  # Replace 'x' with the actual velocity column name

    data.columns = data.columns.str.strip()

    # Calculate terminal velocity and error
    terminal_velocity = velocity.iloc[-10:].mean()  # Average of the last 10 readings
    terminal_velocity_error = velocity.iloc[-10:].std()  # Standard deviation of the last 10 readings

    # Store results in the dictionary
    terminal_velocities[sheet] = {
        'Terminal Velocity': terminal_velocity,
        'Error': terminal_velocity_error
    }

    # Print the results for each sheet
    print(f"Sheet: {sheet}")
    print(f"Terminal Velocity: {terminal_velocity:.4f} m/s")
    print(f"Error: {terminal_velocity_error:.4f} m/s\n")

# Clear and detailed summary printing
print("Summary of Terminal Velocities:")
print("-" * 50)
for sheet, results in terminal_velocities.items():
    print(f"Sheet: {sheet}")
    print(f"  Terminal Velocity: {results['Terminal Velocity']:.4f} m/s")
    print(f"  Error: {results['Error']:.4f} m/s")
    print("-" * 50)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

file_name="tracker_data.xlsx"
data=pd.read_excel(file_name,sheet_name="summary",header=0)
data

# Load the data from the Excel file
# data = pd.read_excel('path_to_your_file.xlsx')

# Extract the relevant columns
radii = data['radius']
radii_error = data['r_e']
terminal_v = data['terminal_v']
v_T_error = data['v_T_e']

# Calculate r^2 for the plot
radii_squared = radii**2

# Plot the terminal velocity vs radius^2 with error bars
plt.errorbar(radii_squared, terminal_v, xerr=radii_error*2*radii, yerr=v_T_error, fmt='o', label='Data with errors')

# Add labels and title
plt.xlabel('Radius squared (m^2)')
plt.ylabel('Terminal Velocity (m/s)')
plt.title('Terminal Velocity vs Radius Squared')

# Fit a linear regression to the data (assuming linear relationship)
slope, intercept, r_value, p_value, std_err = stats.linregress(radii_squared, terminal_v)

# Plot the best fit line
plt.plot(radii_squared, slope * radii_squared + intercept, color='red', label='Fit line')

# Show the plot with legend
plt.legend()
plt.grid(True)
plt.show()

# Calculate viscosity (eta) using the slope and its uncertainty
# Assuming rho_oil - rho_ball = 1000 (a typical value for water)
# g = 9.81 m/s^2, and rho_oil_error is negligible in our calculations.

rho_oil = 900  # Density of oil (kg/m^3)
rho_ball = 1400  # Density of the ball (kg/m^3), example value
g = 9.81  # Acceleration due to gravity (m/s^2)

# Calculate viscosity (eta) from the slope
eta = (2 * (rho_oil - rho_ball) * g) / (9 * slope)

# Propagate the error in slope to the viscosity error
eta_error = eta * std_err / slope

# Print the slope, slope error, viscosity, and viscosity error
print(f"Slope: {slope:.4f} ± {std_err:.4f}")
print(f"Calculated viscosity: {eta:.4f} Pa.s ± {eta_error:.4f}")
```
